# Feature Engineering

This notebook creates features for demand forecasting models.

## Features created:
1. Temporal features (year, month, day of week, etc.)
2. Cyclical encoding for periodic features
3. Lag features
4. Rolling window features
5. Discount-related features


```python
import sys
import os
sys.path.append(os.path.join(os.path.dirname(os.getcwd()), 'src'))

import pandas as pd
import numpy as np
from utils import load_data, save_data
from feature_engineering import prepare_features

# Load cleaned data
data_path = '../data/cleaned/grocery_cleaned.csv'
df = load_data(data_path)

print(f"Original shape: {df.shape}")
print(f"Columns: {df.columns.tolist()}")
```

```python
# Create all features
df_features = prepare_features(
    df, 
    target_col='sales',
    date_col='date',
    include_lags=True,
    include_rolling=True
)

print(f"Features shape: {df_features.shape}")
print(f"\nNew columns created: {len(df_features.columns) - len(df.columns)}")
print(f"\nAll columns: {df_features.columns.tolist()}")
```

```python
# Save features
output_path = '../data/processed/grocery_features.csv'
save_data(df_features, output_path)
print(f"Features saved to {output_path}")
```
