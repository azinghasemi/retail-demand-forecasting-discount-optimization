# Model Training

This notebook trains and compares multiple models for demand forecasting.

## Models:
1. Linear Regression
2. Random Forest
3. XGBoost


```python
import sys
import os
sys.path.append(os.path.join(os.path.dirname(os.getcwd()), 'src'))

import pandas as pd
import numpy as np
from utils import load_data
from model_linear_regression import train_linear_regression, predict_linear_regression, get_feature_importance as lr_importance
from model_random_forest import train_random_forest, predict_random_forest, get_feature_importance as rf_importance
from model_xgboost import train_xgboost, predict_xgboost, get_feature_importance as xgb_importance
from evaluation import print_evaluation_report, compare_models

# Load features
data_path = '../data/processed/grocery_features.csv'
df = load_data(data_path)

# Prepare features and target
target_col = 'sales'
feature_cols = [col for col in df.columns if col not in [target_col, 'date']]
X = df[feature_cols]
y = df[target_col]

print(f"Features: {len(feature_cols)}")
print(f"Samples: {len(X)}")
```

```python
# Train Linear Regression
lr_model, X_train_lr, X_test_lr, y_train_lr, y_test_lr, scaler = train_linear_regression(X, y)
lr_pred = predict_linear_regression(lr_model, X_test_lr, scaler)
lr_metrics = print_evaluation_report(y_test_lr, lr_pred, "Linear Regression")
```

```python
# Train Random Forest
rf_model, X_train_rf, X_test_rf, y_train_rf, y_test_rf = train_random_forest(X, y)
rf_pred = predict_random_forest(rf_model, X_test_rf)
rf_metrics = print_evaluation_report(y_test_rf, rf_pred, "Random Forest")
```

```python
# Train XGBoost
xgb_model, X_train_xgb, X_test_xgb, y_train_xgb, y_test_xgb = train_xgboost(X, y)
xgb_pred = predict_xgboost(xgb_model, X_test_xgb)
xgb_metrics = print_evaluation_report(y_test_xgb, xgb_pred, "XGBoost")
```

```python
# Compare models
models_results = {
    'Linear Regression': lr_metrics,
    'Random Forest': rf_metrics,
    'XGBoost': xgb_metrics
}

comparison_df = compare_models(models_results)
print("\nModel Comparison:")
print(comparison_df)
```
