# Data Cleaning

This notebook performs data cleaning and preprocessing on the raw retail sales data.

## Steps:
1. Load raw data
2. Handle missing values
3. Remove duplicates
4. Handle outliers
5. Validate cleaned data
6. Save cleaned data


```python
import sys
import os
sys.path.append(os.path.join(os.path.dirname(os.getcwd()), 'src'))

import pandas as pd
import numpy as np
from data_prep import clean_retail_data, validate_data
from utils import load_data, save_data

# Load raw data
raw_data_path = '../data/raw/retail_sales_2022_2024_raw.csv'
df_raw = load_data(raw_data_path)

# Display basic info
print("Raw Data Info:")
print(df_raw.info())
print("\nFirst few rows:")
print(df_raw.head())
```

```python
# Clean the data
df_cleaned = clean_retail_data(df_raw)

# Validate cleaned data
is_valid = validate_data(df_cleaned)

if is_valid:
    print("\nData cleaning completed successfully!")
    print(f"Shape: {df_cleaned.shape}")
    print(f"\nSummary statistics:")
    print(df_cleaned.describe())
```

```python
# Save cleaned data
output_path = '../data/cleaned/grocery_cleaned.csv'
save_data(df_cleaned, output_path)
print(f"\nCleaned data saved to {output_path}")
```
